```python
df["Bonus"] = df["Salary"] * 0.10

print(df)
```

```python
df["Total Salary"] = df["Salary"] + df["Bonus"]

print(df)
```

```python
print("Average Salary:", df["Salary"].mean())

print("Maximum Salary:", df["Salary"].max())

print("Minimum Salary:", df["Salary"].min())

print("Total Salary:", df["Salary"].sum())

print("Employee Count:", df["Salary"].count())
```

```python
# Highest-paid employee
highest = df.loc[df["Total Salary"].idxmax()]
print("Highest-paid employee:")
print(f"{highest['Name']} - {highest['Total Salary']}")

# Lowest-paid employee
lowest = df.loc[df["Total Salary"].idxmin()]
print("\nLowest-paid employee:")
print(f"{lowest['Name']} - {lowest['Total Salary']}")

# Average salary
avg_salary = df["Salary"].mean()
print("\nAverage Salary:", avg_salary)

# Employees in the IT department
print("\nEmployees in the IT Department:")
print(df[df["Department"] == "IT"][["ID","Name"]])

# Total salary paid by the company
total_salary = df["Total Salary"].sum()
print("\nTotal Salary Paid by the Company:", total_salary)
```
