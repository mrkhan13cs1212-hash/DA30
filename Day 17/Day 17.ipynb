```python
df["Rank"] = df["Marks"].rank(
    ascending=False,
    method="min"
)
```

```python
df.sort_values("Rank")
```

```python
df[
    (df["Marks"] >= 80) &
    (df["Attendance"] >= 85)
].sort_values("Marks", ascending=False)
```

```python
df.sort_values(
    ["Attendance", "Marks"],
    ascending=[False, False]
).head(3)
```

```python
df.loc[df.groupby("Gender")["Marks"].idxmax()]
```

```python
selected_columns = ["Rank", "Name", "Gender", "Marks", "Attendance"]

top_performers = df[
    (df["Marks"] >= 75) &
    (df["Attendance"] >= 80)
].sort_values("Marks", ascending=False)

print(top_performers[selected_columns])
```

```python
df.loc[df.groupby("Gender")["Marks"].idxmax()]
```

```python
selected_columns = ["Name", "Gender", "Marks", "Attendance", "Rank"]

top_by_gender = df.loc[
    df.groupby("Gender")["Marks"].idxmax(),
    selected_columns
]

print(top_by_gender)
```

```python
selected_columns = ["Name", "Gender", "Marks", "Attendance", "Rank"]

top_by_gender = df.loc[
    df.groupby("Department")["Marks"].idxmax(),
    selected_columns
]

print(top_by_gender)
```
