```python
#Challenge
import numpy as np

sales = np.array([
    1200, 900, 1800, 1500,
    2200, 800, 1700, 1100,
    1900, 2100, 1300, 1600
])

print("Original Array:",sales)
print("Array Shape:",sales.shape)

reshape = sales.reshape(3, 4)

print(reshape)
print("Reshape Size", reshape.shape)
print("Highest: ", np.max(sales))
print("Lowest: ", np.min(sales))
print("Assending order: ", np.sort(sales))
print("Decending order: ", np.sort(sales)[::-1])
print("Average: ", np.average(sales))
print("Top 3: ", np.sort(sales)[-3:])
print("Least 3: ", np.sort(sales)[:3])
```
