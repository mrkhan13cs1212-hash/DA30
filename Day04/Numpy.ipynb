```python
import numpy as np

marks = np.array([85,72,91,67,88,95,78,69,82,90])

print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Total:", np.sum(marks))

print("Students above 80:")
print(marks[marks>80])

print("No.of Students passed: ", np.count_nonzero(marks[marks>35]))
print("No.of Students above 90: ", np.count_nonzero(marks[marks>90]))
print("Average percentage: ", (np.mean(marks)/100)*100,"%")
```
