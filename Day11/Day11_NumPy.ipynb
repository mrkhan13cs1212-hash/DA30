```python
import numpy as np

sales = np.array([
    [1200, 1500, 1800, 1400, 2000],
    [900, 1100, 1300, 1250, 1600],
    [2200, 2100, 2300, 2500, 2400]
])

print("Sales:", sales)
```

```python
print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Total Sales of Stores:",np.sum(sales, axis=1))
print("Average Sales of Stores:",np.mean(sales, axis=1))
print("Total Sales of Monday:",np.sum(sales[:, 0]))
print("Total Sales of Tuesday:",np.sum(sales[:, 1]))
print("Total Sales of Wednesday:",np.sum(sales[:, 2]))
print("Total Sales of Thursday:",np.sum(sales[:, 3]))
print("Total Sales of Friday:",np.sum(sales[:, 4]))
print("Average Sales of Monday:",np.mean(sales[:, 0]))
print("Average Sales of Tuesday:",np.mean(sales[:, 1]))
print("Average Sales of Wednesday:",np.mean(sales[:, 2]))
print("Average Sales of Thursday:",np.mean(sales[:, 3]))
print("Average Sales of Friday:",np.mean(sales[:, 4]))
print("Index of Highest Sales:", np.unravel_index(np.argmax(sales), sales.shape))
print("Maximum sale:",np.max(sales))
print("Minimum sale:",np.min(sales))
```
