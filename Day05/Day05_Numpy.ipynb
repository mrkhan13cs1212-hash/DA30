```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)

marks = np.array([50, 60, 70])

print(marks + 5)
print(marks * 2)

marks = np.array([25, 40, 55, 70, 85])

print(marks > 50)
print(marks[marks > 50])
```

```python
marks = np.array([25, 40, 55, 70, 85])

print(marks[(marks > 40) & (marks < 80)])

print(np.random.randint(1, 100, 10))
```

```python
marks = np.array([45, 67, 82, 90, 33, 76, 58])

print(marks[marks>60])
print(marks[marks<35])
print(np.count_nonzero(marks>75))
print(len(marks[marks > 75]))
```

```python
sales = np.array([1200, 1500, 1800, 900, 2000, 1750, 800])

print("Total Weekly Sale: ")
print(np.sum(sales))
print("Average daily sale:")
print(np.mean(sales))
print("Highest and Lowest Sales:")
print(np.max(sales),"&", np.min(sales))
print("Days where sales exceeded 1500")
print(sales[sales > 1500])
```
